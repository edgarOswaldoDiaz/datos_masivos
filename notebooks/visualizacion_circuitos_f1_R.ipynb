
# Visualización de Circuitos de Fórmula 1

Este notebook carga el archivo `f1_circuits.json` y visualiza la ubicación geográfica de los circuitos usando **R**.
Cada línea de código está documentada para facilitar su comprensión.


```python

# Cargar biblioteca para leer archivos JSON
library(jsonlite)  # Permite convertir archivos JSON a data frames

# Cargar biblioteca para manipulación de datos
library(dplyr)     # Facilita operaciones como seleccionar y filtrar columnas

# Cargar biblioteca para visualización
library(ggplot2)   # Se usa para crear gráficos avanzados

# Leer el archivo JSON que contiene los circuitos de F1
circuits <- fromJSON('f1_circuits.json')  # Convierte el JSON en un data frame

# Ver las primeras filas del conjunto de datos
head(circuits)     # Muestra una vista previa para verificar la carga correcta

# Crear un mapa mundial simple con los circuitos
world_map <- map_data('world')  # Obtiene datos geográficos del mundo

# Graficar el mapa y los circuitos

ggplot() +
  geom_polygon(data = world_map,               # Dibuja el mapa del mundo
               aes(x = long, y = lat, group = group),
               fill = 'gray90', color = 'white') +
  geom_point(data = circuits,                  # Agrega los circuitos
             aes(x = lng, y = lat),
             color = 'red', size = 2) +
  labs(title = 'Circuitos de Fórmula 1 en el mundo',  # Título del gráfico
       x = 'Longitud', y = 'Latitud') +
  theme_minimal()                               # Aplica un tema limpio
```
