# Visualización del archivo `f1_circuits.csv`

Este Jupyter Notebook tiene como objetivo **mostrar y explicar paso a paso** cómo cargar, explorar y visualizar el contenido del archivo `f1_circuits.csv`.

Cada línea de código está documentada detalladamente para facilitar su comprensión, incluso para personas que están aprendiendo Python, pandas o Jupyter Notebooks.

## 1. Importación de librerías

En esta sección se importan las bibliotecas necesarias para el análisis y visualización de datos.

```python
# Importamos la libreríabiblioteca pandas, que se utiliza para cargar y manipular datos en forma de tablas (DataFrames)
import pandas as pd

# Importamos matplotlib.pyplot, un módulo muy usado para crear gráficos y visualizaciones
import matplotlib.pyplot as plt
```

## 2. Carga del archivo CSV

Ahora se carga el archivo `f1_circuits.csv` en memoria para poder trabajar con sus datos.

```python
# Leemos el archivo CSV y lo almacenamos en una variable llamada 'df' (abreviatura de DataFrame)
# pandas detecta automáticamente las columnas y tipos de datos
df = pd.read_csv('f1_circuits.csv')

# Mostramos las primeras 5 filas del DataFrame para verificar que la carga fue exitosa
df.head()
```

## 3. Exploración básica del dataset

En esta sección se obtiene información general del conjunto de datos, como número de filas, columnas y tipos de datos.

```python
# df.info() muestra un resumen del DataFrame:
# - Número total de filas
# - Nombre de cada columna
# - Tipo de dato de cada columna
# - Cantidad de valores no nulos
df.info()
```

## 4. Visualización gráfica de los circuitos

Utilizando las columnas de latitud y longitud, se crea una gráfica para visualizar la ubicación aproximada de los circuitos de Fórmula 1 en el mundo.

```python
# Creamos una figura para controlar el tamaño del gráfico
plt.figure(figsize=(10, 5))

# Dibujamos un gráfico de dispersión (scatter plot)
# - Eje X: longitud (lng)
# - Eje Y: latitud (lat)
plt.scatter(df['lng'], df['lat'])

# Etiquetamos el eje X
plt.xlabel('Longitud')

# Etiquetamos el eje Y
plt.ylabel('Latitud')

# Agregamos un título descriptivo al gráfico
plt.title('Ubicación geográfica de los circuitos de Fórmula 1')

# Mostramos el gráfico en pantalla
plt.show()
```
